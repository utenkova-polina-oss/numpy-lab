```python
# Create a random array of numbers between 0 & 10 and an array of ones
# both of size (3, 5), save them both to variables
random_array_2 = np.random.randint(0, 10, size=(3, 5))
ones = np.ones((3, 5))
random_array_2, ones
```

```python
# Add the two arrays together
random_array_2 + ones
```

```python
# Create another array of ones of shape (5, 3)
ones_2 = np.ones((5, 3))
ones_2
```

```python
# Try add the array of ones and the other most recent array together
try:
    ones_2 + random_array_2
except ValueError as e:
    print("Error:", e)
```

When you try the last cell, it produces an error. Why do think this is?

How would you fix it?

**Ответ:** ошибка возникает, потому что формы массивов не совпадают: (5, 3) и (3, 5). Поэлементное сложение и broadcasting требуют, чтобы размерности совпадали или одна из них была равна 1. Исправить можно, взяв массив единиц формы (3, 5) или транспонировав один из массивов (`ones_2.T`).

```python
# Create another array of ones of shape (3, 5)
ones_3 = np.ones((3, 5))
ones_3
```

```python
# Subtract the new array of ones from the other most recent array
random_array_3 = random_array_2 - ones_3
random_array_3
```

```python
# Multiply the ones array with the latest array
random_array_4 = ones_3 * random_array_3
random_array_4
```

```python
# Take the latest array to the power of 2 using '**'
random_array_4 ** 2
```

```python
# Transpose the latest array
reshaped.T, reshaped.T.shape
```

What does the transpose do?

**Ответ:** транспонирование меняет оси массива местами в обратном порядке: строки становятся столбцами. Массив формы (3, 5, 1) превращается в массив формы (1, 5, 3).

```python
# Create two arrays of random integers between 0 to 10
# one of size (3, 3) the other of size (3, 2)
mat_1 = np.random.randint(0, 10, size=(3, 3))
mat_2 = np.random.randint(0, 10, size=(3, 2))
mat_1, mat_2
```

```python
# Perform a dot product on the two newest arrays you created
np.dot(mat_1, mat_2)
```

```python
# Create two arrays of random integers between 0 to 10
# both of size (4, 3)
mat_3 = np.random.randint(0, 10, size=(4, 3))
mat_4 = np.random.randint(0, 10, size=(4, 3))
mat_3, mat_4
```

```python
# Perform a dot product on the two newest arrays you created
try:
    np.dot(mat_3, mat_4)
except ValueError as e:
    print("Error:", e)
```

It doesn't work. How would you fix it?

**Ответ:** для скалярного (матричного) произведения число столбцов первой матрицы должно совпадать с числом строк второй. У (4, 3) и (4, 3) это не так: 3 ≠ 4. Нужно транспонировать одну из матриц.

```python
# Take the latest two arrays, perform a transpose on one of them and then perform 
# a dot product on them both
np.dot(mat_3, mat_4.T)
```

Notice how performing a transpose allows the dot product to happen.

Why is this?

Checking out the documentation on [`np.dot()`](https://docs.scipy.org/doc/numpy/reference/generated/numpy.dot.html) may help, as well as reading [Math is Fun's guide on the dot product](https://www.mathsisfun.com/algebra/vectors-dot-product.html).

Let's now compare arrays.

**Ответ:** после транспонирования формы становятся (4, 3) и (3, 4): внутренние размерности совпадают (3 = 3), и произведение возможно. Результат имеет форму внешних размерностей — (4, 4).

```python
# Create two arrays of random integers between 0 & 10 of the same shape
# and save them to variables
nums_1 = np.random.randint(0, 10, size=(3, 4))
nums_2 = np.random.randint(0, 10, size=(3, 4))
nums_1, nums_2
```

```python
# Compare the two arrays with '>'
nums_1 > nums_2
```

What happens when you compare the arrays with `>`?

**Ответ:** сравнение выполняется поэлементно, и результат — массив той же формы из значений `True` / `False`: для каждой позиции указано, больше ли элемент первого массива элемента второго.

```python
# Compare the two arrays with '>='
nums_1 >= nums_2
```

```python
# Find which elements of the first array are greater than 7
nums_1 > 7
```

```python
# Which parts of each array are equal? (try using '==')
nums_1 == nums_2
```

```python
# Sort one of the arrays you just created in ascending order
np.sort(nums_1)
```

```python
# Find the index with the maximum value in one of the arrays you've created
np.argmax(nums_1)
```

```python
# Find the index with the minimum value in one of the arrays you've created
np.argmin(nums_1)
```

```python
# Find the indexes with the maximum values down the 1st axis (axis=1)
# of one of the arrays you created
np.argmax(nums_1, axis=1)
```

```python
# Find the indexes with the minimum values across the 0th axis (axis=0)
# of one of the arrays you created
np.argmin(nums_1, axis=0)
```
